**1. Create the Student Database**

```r
students <- data.frame(
    Student_ID = c("S001", "S002", "S003", "S004", "S005", "S006", "S007", "S008"),
    Programming_Grade = c(88, 72, 91, 78, 65, 85, 74, 82),
    Math_Grade = c(85, 70, 88, 80, 68, 90, 73, 79),
    Attendance = c(92, 80, 95, 88, 70, 91, 76, 85),
    Prerequisite = c("Yes","No")
)

students
```

**2. Explore the Data**

```r
nrow(students)
ncol(students)
max(students$Programming_Grade)
max(students$Attendance)
max(students$Math_Grade)
mean(students$Attendance)
min(students$Math_Grade)
```

**6. Extract Student Information**

```r
programming <- student$Programming_Grade
math <- student$Math_Grade
attendance <- student$Attendance
prereq <- student$Prerequisite

cat("Programming Grade:", programming)
cat("\nMath Grade:", math)
cat("\nAttendance:", attendance, "%")
cat("\nPrerequisite:", prereq)
```

**7. Implement the Rules**

```r
if (programming >= 80 &&
    math >= 75 &&
    attendance >= 80 &&
    prereq == "Yes") {

  classification <- "Eligible"

} else if (programming >= 75 &&
           math >= 70 &&
           attendance >= 75) {

  classification <- "Conditionally Eligible"

} else {

  classification <- "Not Eligible"
}

cat("\nClassification:", classification)
```

**8. Apply the Function to All Students**

```r
check_eligibility <- function(programming, math, attendance, prereq) {

  if (programming >= 80 &&
      math >= 75 &&
      attendance >= 80 &&
      prereq == "Yes") {

    return("Eligible")

  } else if (programming >= 75 &&
             math >= 70 &&
             attendance >= 75) {

    return("Conditionally Eligible")

  } else {

    return("Not Eligible")
  }
}

students$Classification <- mapply(
  check_eligibility,
  students$Programming_Grade,
  students$Math_Grade,
  students$Attendance,
  students$Prerequisite
)

students
```

**9. Create Recommendation Rules**

```r
recommendation <- function(programming, math, attendance, prereq) {

  if (programming >= 80 &&
      math >= 75 &&
      attendance >= 80 &&
      prereq == "Yes") {

    return("Highly Recommended")

  } else if (programming >= 75 &&
             math >= 70 &&
             attendance >= 75) {

    return("Recommended with Improvement")

  } else {

    return("Needs Improvement")
  }
}

students$Recommendation <- mapply(
  recommendation,
  students$Programming_Grade,
  students$Math_Grade,
  students$Attendance,
  students$Prerequisite
)

students
```

**Final Output**

```r
# ASK FOR STUDENT ID
student_id <- readline("Enter Student ID: ")

# FIND STUDENT
student <- students[students$Student_ID == student_id, ]

if (nrow(student) == 0) {

  cat("\nStudent not found.\n")

} else {

  # EXTRACT STUDENT INFORMATION
  programming <- student$Programming_Grade
  math <- student$Math_Grade
  attendance <- student$Attendance
  prereq <- student$Prerequisite

  # IMPLEMENT RULES
  if (programming >= 80 &&
      math >= 75 &&
      attendance >= 80 &&
      prereq == "Yes") {

    classification <- "Eligible"

  } else if (programming >= 75 &&
             math >= 70 &&
             attendance >= 75) {

    classification <- "Conditionally Eligible"

  } else {

    classification <- "Not Eligible"
  }

  # RECOMMENDATION RULES
  if (classification == "Eligible") {

    recommendation <- "Proceed with enrollment"

  } else if (classification == "Conditionally Eligible") {

    recommendation <- "Consult academic adviser"

  } else {

    recommendation <- "Not recommended for enrollment"
  }

  # FINAL OUTPUT
  cat("\n==========================================\n")
  cat("          COURSE ENROLLMENT ADVISER\n")
  cat("==========================================\n\n")

  cat("Student ID:", student$Student_ID, "\n\n")

  cat("Programming Grade:", programming, "\n")
  cat("Math Grade:", math, "\n")
  cat("Attendance:", attendance, "%\n")
  cat("Prerequisite:", prereq, "\n\n")

  cat("Enrollment Status:\n")
  cat(classification, "\n\n")

  cat("Recommendation:\n")
  cat(recommendation, "\n\n")

  cat("==========================================\n")
}
```
